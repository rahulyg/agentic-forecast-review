# Credentials

Find the OpenRouter API key: first the `OPENROUTER_API_KEY` environment variable, then
a `.env` file in the repo root (ignored by git). The placeholder value counts as missing.

There is no example here on purpose: running it would print your real key.

```python
#| default_exp credentials
```

```python
#| export
import os
from pathlib import Path

ENV_VAR = "OPENROUTER_API_KEY"
PLACEHOLDER_KEY = "replace_with_your_openrouter_api_key"
```

Return the key, or raise if it is missing or still the placeholder.

```python
#| export
def openrouter_api_key() -> str:
    key = os.environ.get(ENV_VAR, "").strip()
    if key and key != PLACEHOLDER_KEY:
        return key
    env_path = Path.cwd() / ".env"
    if env_path.is_file():
        for line in env_path.read_text(encoding="utf-8").splitlines():
            name, separator, value = line.strip().partition("=")
            if name == ENV_VAR and separator:
                key = value.strip()
                break
    if not key or key == PLACEHOLDER_KEY:
        raise RuntimeError(
            "OpenRouter API key is missing or a placeholder; set OPENROUTER_API_KEY."
        )
    return key
```
